```python
# string concatenation
str1 = "hello "
str2 = "kevin"
result = str1 + str2
print(result)
```

```python
str3 = " welcome to python programming"
result3 = result + str3
print(result3)
```

```python
# string indexing and slicing
text = "hello kevin welcome to python programming"
print("first character:",text[0])
```

```python
print("last character:",text[-1])
```

```python
print("first 5 character:",text[:5])
```

```python
print("last 11 character:",text[-11:])
```

```python
print("reversed string:",text[::-1])
```

```python
print("reversed string:",text[0::])
```

```python
print("word python:",text[23:29])
```

```python
# string methods
strM = "Python beginner tutorial"
print("uppercase:",strM.upper())
print("lowercase:",strM.lower())
print("capitalized:",strM.capitalize())
print("occurrences of 't':",strM.count('t'))
print("replaced string:",strM.replace("python","machine learning"))
```

```python
#tuples
t1 = (10,20,30)
t2 = (40,50,60)
t_combine = t1 + t2
print("combined tuples:",t_combine)
```

```python
print("repeated tuples:",t_combine*3)
```

```python
print("third element:",t_combine[2])
```

```python
print("first 3 elements:",t_combine[:3])
```

```python
print("last 3 elements:",t_combine[-3:])
```
